# Market data and sessions

A four-row deterministic timeline demonstrates that session labeling retains rows. No external data is used, and this output is not trading evidence.

```python
import pandas as pd

from market_data.transforms.sessions import add_session_labels

idx = pd.date_range("2026-01-05 09:00", periods=4, freq="1h", tz="Etc/GMT-3")
bars = pd.DataFrame({"open": [1.0, 2.0, 3.0, 4.0]}, index=idx)
labeled = add_session_labels(bars)
assert labeled.index.equals(bars.index)
assert len(labeled) == len(bars)
labeled[["session"]]
```
