# Walk-forward splits

A deterministic index range demonstrates disjoint test blocks and purge plus embargo gaps. No external data is used, and this output is not trading evidence.

```python
from market_data.splits.walkforward import purged_walk_forward

folds = list(
    purged_walk_forward(
        1000, n_splits=3, test_size=0.3, purge_bars=10, embargo_bars=5, label_horizon=7
    )
)
assert len(folds) == 3
for left, right in zip(folds, folds[1:]):
    assert left.test_idx[-1] < right.test_idx[0]
for fold in folds:
    assert fold.train_idx[-1] + 10 + 7 + 5 < fold.test_idx[0]
[
    (int(f.train_idx[0]), int(f.train_idx[-1]), int(f.test_idx[0]), int(f.test_idx[-1]))
    for f in folds
]
```
