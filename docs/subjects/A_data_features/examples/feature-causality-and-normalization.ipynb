# Feature causality and normalization

A tiny higher-timeframe frame demonstrates that completed values do not appear early. No external data is used, and this output is not trading evidence.

```python
import pandas as pd

from market_data.transforms.align import align_timeframes

m1_idx = pd.date_range("2026-01-05 09:00", periods=8, freq="1min", tz="Etc/GMT-3")
h1_idx = pd.date_range("2026-01-05 09:00", periods=2, freq="4min", tz="Etc/GMT-3")
m1 = pd.DataFrame({"close": range(8)}, index=m1_idx)
higher = pd.DataFrame({"close": [100.0, 200.0]}, index=h1_idx)
aligned = align_timeframes(m1, {"H1": higher})
assert aligned["H1_close"].iloc[:4].isna().all()
assert aligned["H1_close"].iloc[4] == 100.0
assert aligned["H1_close"].iloc[7] == 100.0
aligned[["H1_close"]]
```
