# Ticks and resampling

Ten synthetic M1 bars demonstrate left-labeled M5 aggregation and summed tick activity. No external data is used, and this output is not trading evidence.

```python
import pandas as pd

from market_data.transforms.resample import resample

idx = pd.date_range("2026-01-05 09:00", periods=10, freq="1min", tz="Etc/GMT-3")
m1 = pd.DataFrame(
    {
        "open": range(10),
        "high": range(1, 11),
        "low": range(10),
        "close": range(1, 11),
        "tickvol": [2] * 10,
    },
    index=idx,
)
m5 = resample(m1, "M5")
assert len(m5) == 2
assert m5.iloc[0]["open"] == 0
assert m5.iloc[0]["high"] == 5
assert m5.iloc[0]["low"] == 0
assert m5.iloc[0]["close"] == 5
assert m5.iloc[0]["tickvol"] == 10
m5
```
