# Equity reward and component attribution

Deterministic CPU contract example with no market data or performance claim.

```python
from math import isclose

from simulation.environment.rewards import REWARD_PART_KEYS, EquityDeltaReward

path = (100_000.0, 100_125.0, 99_950.0, 100_400.0)
reward = EquityDeltaReward(500.0)
steps = []
for left, right in zip(path[:-1], path[1:], strict=False):
    value = reward(right - left)
    assert isclose(reward.last_parts["pnl"], value)
    assert isclose(sum(reward.last_parts.values()), value)
    steps.append(value)
assert isclose(sum(steps), (path[-1] - path[0]) / 500.0)
assert all(key in reward.last_parts for key in REWARD_PART_KEYS)
assert reward(500.0, breach=True) == -10.0
assert reward.last_parts["breach"] == -10.0
print(
    {"step_rewards": steps, "path_sum": sum(steps), "breach_component": reward.last_parts["breach"]}
)
```
