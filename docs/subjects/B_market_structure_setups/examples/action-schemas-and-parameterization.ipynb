# Action schemas and parameterization

This schema-only example compares same-width layouts and prints a setup-conditioned layout. It uses no market data and makes no policy-quality claim.

```python
from decisioning.actions import build_action_schema

direction = build_action_schema(has_direction=True)
sl_mode = build_action_schema(has_sl_mode=True)
assert direction.width == sl_mode.width
assert direction.names != sl_mode.names
assert direction.key != sl_mode.key
print("same width:", direction.width)
print("direction:", direction.names)
print("stop mode:", sl_mode.names)
```

```python
from decisioning.trade_setup import build_trade_setup_action_schema

schema = build_trade_setup_action_schema(
    has_multi_tp=True,
    has_simplex=True,
    has_entry_execution=True,
)
print(schema.to_dict())
assert schema.version == "trade-setup-entry-execution-action-v2"
assert "entry_execution" in schema.names
assert "limit_offset_atr" in schema.names
assert "limit_ttl_bars" in schema.names
```
