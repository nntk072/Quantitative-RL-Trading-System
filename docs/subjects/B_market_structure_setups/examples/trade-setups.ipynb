# Candidate IDs and exact eligibility mask

This deterministic CPU example illustrates a software contract only. It uses no market feed and provides no evidence of trading performance.

```python
import numpy as np

from decisioning.trade_setup import (
    TRADE_SETUP_CATEGORY_COUNT,
    trade_setup_candidate_id,
    trade_setup_center,
    trade_setup_id_from_box,
)
from features.context.trade_setup_candidates import trade_setup_candidate_mask

assert TRADE_SETUP_CATEGORY_COUNT == 109
candidate_id = trade_setup_candidate_id("fvg_retest", "M5", 1)
encoded = trade_setup_center(candidate_id)
assert trade_setup_id_from_box(encoded) == candidate_id
mask = trade_setup_candidate_mask({"entry_idea_fvg_retest_M5_long": 1.0})
assert mask.shape == (TRADE_SETUP_CATEGORY_COUNT,)
assert mask[0] == 1.0
assert mask[candidate_id] == 1.0
assert np.count_nonzero(mask) == 2
(candidate_id, encoded, trade_setup_id_from_box(encoded), np.flatnonzero(mask))
```
