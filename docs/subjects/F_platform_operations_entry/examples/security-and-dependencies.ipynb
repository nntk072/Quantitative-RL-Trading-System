# Read advisory findings and audit coverage

The bundled payload is synthetic. It represents an advisory finding and a package that the scanner skipped. Parse it locally with the audit wrapper's report helper.

```python
from pathlib import Path
from runpy import run_path

audit_file = Path("scripts/dev/security/dependency_audit.py")
start = Path.cwd()
checkout = next(
    directory for directory in (start, *start.parents) if (directory / audit_file).is_file()
)
parse_report = run_path(str(checkout / audit_file))["findings_from_report"]

payload = {
    "dependencies": [
        {"name": "Example_Pkg", "version": "1.0", "vulns": [{"id": "demo-advisory"}]},
        {"name": "local-wheel", "version": "1.0", "skip_reason": "missing index metadata"},
    ]
}
findings, skips = parse_report(payload)
assert findings == [("example-pkg", "DEMO-ADVISORY", "1.0")]
assert skips == ["local-wheel@1.0: missing index metadata"]
{"findings": findings, "skips": skips}
```
