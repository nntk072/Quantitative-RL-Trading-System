# Read a broker-neutral signal and its configured risk cap

The fixture supplies one directional signal and a synthetic saved training budget. Signal typing and risk configuration can be inspected offline. The example has no broker connection.

```python
from integrations.mt5.risk_configuration import load_live_risk_configuration
from live.runtime.contracts import LiveSignal, LiveStrategy


class FixedStrategy:
    def signal(self) -> tuple[str, LiveSignal]:
        return "SYNTHETIC", LiveSignal.HOLD


strategy: LiveStrategy = FixedStrategy()
symbol, signal = strategy.signal()
fixture = {
    "ftmo": {"risk_sizing_limit_usd": 120.0},
    "live_risk_overrides": {
        "risk_per_symbol": 0.01,
        "max_total_risk": 0.02,
        "stop_loss_pips": 50.0,
        "interval_minutes": 1.0,
    },
    "costs": {"slippage_points": 0.0, "commission_per_lot": 0.0},
}
risk = load_live_risk_configuration(fixture)
assert signal is LiveSignal.HOLD
assert risk.risk_per_symbol_usd_cap == 120.0
{"symbol": symbol, "signal": signal.value, "stop_risk_cap_usd": risk.risk_per_symbol_usd_cap}
```
