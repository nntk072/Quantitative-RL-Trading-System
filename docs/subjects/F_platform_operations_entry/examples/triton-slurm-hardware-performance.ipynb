# Find the limiting resource in a worker plan

This synthetic node record describes one allocation with fixed CPU IDs, host memory, and a GPU memory estimate. The planner only reads the record. No worker starts.

```python
from compute.slurm.worker_resources import plan_worker_capacity

gib = 1024**3
node_resources = {
    "fixture-node": {
        "cpu_ids": list(range(32)),
        "memory_limit_bytes": 120 * gib,
        "gpu_memory": [{"total_bytes": 96 * gib, "free_bytes": 80 * gib}],
    }
}
plan = plan_worker_capacity(
    node_resources,
    n_envs=8,
    data_source="frames",
    gpus_per_node=1,
)
node = plan["nodes"]["fixture-node"]
assert node["cpu_capacity"] == 7
assert node["memory_capacity"] == 2
assert node["gpu_capacity"] == 9
assert plan["workers"] == node["workers"] == 2
{key: node[key] for key in ("cpu_capacity", "memory_capacity", "gpu_capacity", "workers")}
```
