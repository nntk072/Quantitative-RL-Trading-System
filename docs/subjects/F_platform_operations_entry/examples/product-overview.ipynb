# Find the installed workflow entry points

The bundled expected command map records the entry-point names and their targets. Compare it with the installed distribution's metadata without importing or running any commands.

```python
from importlib.metadata import distribution

expected = {
    "qrl-data": "market_data.cli:main",
    "qrl-train": "training.cli:main",
    "qrl-evaluate": "evaluation.cli:main",
    "qrl-experiment": "experiments.cli:main",
    "qrl-compute": "compute.cli:main",
    "qrl-study": "research_tools.studies.study_cli:main",
    "orchestra": "orchestra.cli:main",
}
project = distribution("quantitative-rl-trading-system")
routes = {
    point.name: point.value for point in project.entry_points if point.group == "console_scripts"
}
assert routes == expected
dict(sorted(routes.items()))
```
