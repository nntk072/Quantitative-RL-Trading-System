# PPO and rollout geometry: deterministic example

The notebook calculates effective samples for a small configured budget and compares full and partial minibatches. It does not instantiate SB3 or train a policy.

Synthetic values only; no research result.

```python
configured_steps = 10
n_envs = 3
per_env_steps = max(1, configured_steps // n_envs)
samples = per_env_steps * n_envs
batch_size = 4
batches = [min(batch_size, samples - start) for start in range(0, samples, batch_size)]
assert (per_env_steps, samples, batches) == (3, 9, [4, 4, 1])
print(f"per environment: {per_env_steps}; aggregate: {samples}; minibatches: {batches}")
```

The assertion checks this bundled example. It does not validate a learned policy, market data, or economic outcome.
