# Training seeds, resume, and checkpoints: deterministic example

The notebook computes remaining steps and a SHA-256 digest of bundled bytes, then rejects an over-budget checkpoint in a small helper. It creates no file and resumes no training.

Synthetic values only; no research result.

```python
from hashlib import sha256

requested_total = 100
completed = 40
remaining = requested_total - completed
assert remaining == 60
checkpoint_bytes = b"synthetic checkpoint example v1"
digest = sha256(checkpoint_bytes).hexdigest()
assert len(digest) == 64
try:
    if requested_total < 120:
        raise ValueError("checkpoint steps exceed requested total")
except ValueError as error:
    print("Rejected:", error)
print("Remaining steps:", remaining, "example digest:", digest[:12])
```

The assertion checks this bundled example. It does not validate a learned policy, market data, or economic outcome.
