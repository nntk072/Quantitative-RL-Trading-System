# Neural encoders and multitimeframe observations: deterministic example

The notebook demonstrates mask-aware extraction of the final valid row from fixed M1 and M5 arrays. It does not instantiate Torch or claim the numeric output equals a trained encoder.

Synthetic values only; no research result.

```python
streams = {
    "m1": {"values": [0.0, 1.0, 2.0], "mask": [0, 1, 1]},
    "m5": {"values": [0.0, 0.0, 8.0], "mask": [0, 0, 1]},
}
last_valid = {}
for name, branch in streams.items():
    valid = [value for value, mask in zip(branch["values"], branch["mask"], strict=True) if mask]
    last_valid[name] = valid[-1] if valid else 0.0
assert last_valid == {"m1": 2.0, "m5": 8.0}
print("Last valid rows:", last_valid)
```

The assertion checks this bundled example. It does not validate a learned policy, market data, or economic outcome.
