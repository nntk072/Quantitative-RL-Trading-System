# Constrained RL: deterministic example

The notebook applies the exact bounded dual-update equation to a fixed sequence of completed episode costs. It shows a zero-cost rollout with no update separately.

Synthetic values only; no research result.

```python
budget = 0.5
rate = 0.2
lambda_max = 2.0
multiplier = 0.0
completed_episode_means = [0.75, 1.0, 0.25]
trajectory = []
for observed in completed_episode_means:
    multiplier = min(lambda_max, max(0.0, multiplier + rate * (observed - budget)))
    trajectory.append(round(multiplier, 4))
assert trajectory == [0.05, 0.15, 0.1]
print("Multiplier after completed episodes:", trajectory)
```

The assertion checks this bundled example. It does not validate a learned policy, market data, or economic outcome.
