# MDP and POMDP: deterministic example

Two hidden states emit the same visible symbol; a short observation history separates only some histories. The notebook computes a small belief update with fixed probabilities and no market data.

Synthetic values only; no research result.

```python
from fractions import Fraction

# Hidden states calm/volatile may both emit the same observed value.
prior = {"calm": Fraction(3, 4), "volatile": Fraction(1, 4)}
likelihood = {"calm": Fraction(1, 2), "volatile": Fraction(3, 4)}
weights = {state: prior[state] * likelihood[state] for state in prior}
normalizer = sum(weights.values())
posterior = {state: weight / normalizer for state, weight in weights.items()}
assert posterior == {"calm": Fraction(2, 3), "volatile": Fraction(1, 3)}
print("Posterior after one ambiguous observation:", posterior)
```

The assertion checks this bundled example. It does not validate a learned policy, market data, or economic outcome.
