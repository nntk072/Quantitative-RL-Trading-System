# Soft Actor-Critic (SAC): deterministic example

The notebook keeps a three-transition replay list and computes a fixed weighted reward example. It illustrates reuse of stored samples only; it is not an SAC implementation.

Synthetic values only; no research result.

```python
transitions = [
    {"reward": 1.0, "done": False},
    {"reward": -0.5, "done": False},
    {"reward": 0.25, "done": True},
]
indices = [0, 2, 0]  # A stored transition can be selected again.
replay_rewards = [transitions[index]["reward"] for index in indices]
assert replay_rewards == [1.0, 0.25, 1.0]
print("Fixed replay sample:", replay_rewards)
```

The assertion checks this bundled example. It does not validate a learned policy, market data, or economic outcome.
