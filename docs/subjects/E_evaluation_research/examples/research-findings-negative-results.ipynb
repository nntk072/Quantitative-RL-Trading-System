# Research findings and negative results: deterministic example

Preserve each synthetic path's observed status and check the implications for a gate requiring every path to complete without a hard breach.

```python
planned = {"seed-1", "seed-2", "seed-3"}
observed = {
    "seed-1": {"status": "breached", "terminal_disposition": "hard_breach", "trade_count": 1},
    "seed-2": {"status": "no_trade", "terminal_disposition": "full_path", "trade_count": 0},
}
missing = planned - observed.keys()
statuses = {path: record["status"] for path, record in observed.items()}
statuses.update({path: "not measured" for path in sorted(missing)})

full_coverage = not missing
breach_detected = any(
    record["terminal_disposition"] == "hard_breach" for record in observed.values()
)
all_paths_survived = full_coverage and all(
    record["terminal_disposition"] == "full_path" for record in observed.values()
)
assert statuses == {"seed-1": "breached", "seed-2": "no_trade", "seed-3": "not measured"}
assert breach_detected and not full_coverage and not all_paths_survived
print("Synthetic path statuses:", statuses)
print("Synthetic all-path survival gate passed:", all_paths_survived)
```

The valid breach fails the gate. The missing path limits coverage and has no observed account outcome. This example provides no research finding or reproduction level.
