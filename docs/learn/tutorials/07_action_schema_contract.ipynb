# Named actions and schema order

An action vector is meaningful only with its schema. Here, two same-width layouts differ in named coordinate meaning.


```python
from decisioning.actions import build_action_schema

stop_layout = build_action_schema(has_sl_mode=True)
target_layout = build_action_schema(has_tp_mode=True)
print(stop_layout.names)
print(target_layout.names)
assert stop_layout.width == target_layout.width
assert stop_layout.names != target_layout.names
assert stop_layout.key != target_layout.key
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 2.5))
for y, schema in enumerate((stop_layout, target_layout)):
    ax.scatter(range(schema.width), [y] * schema.width, s=260)
    for x, name in enumerate(schema.names):
        ax.text(x, y, name, ha="center", va="center", fontsize=7, color="white")
ax.set(
    yticks=[0, 1],
    yticklabels=[stop_layout.key, target_layout.key],
    xlabel="Action coordinate",
    title="Same vector width, different schema",
)
plt.tight_layout()
plt.show()
```

A checkpoint and its consumer must agree on schema identity and coordinate order. Matching vector length alone is insufficient. See the [action schema reference](../../reference/action_schemas.md).
