# Purged walk-forward validation

This notebook visualizes the repository splitter. Test folds are disjoint, and training stops before each test interval by the configured purge, label horizon, and embargo. The fixture is synthetic and demonstrates split geometry only.


```python
from market_data.splits.walkforward import purged_walk_forward

n_rows = 240
folds = list(
    purged_walk_forward(
        n_rows, n_splits=4, test_size=0.2, purge_bars=8, embargo_bars=4, label_horizon=3
    )
)
assert len(folds) == 4
for fold in folds:
    print(fold.fold, (fold.train_idx[0], fold.train_idx[-1]), (fold.test_idx[0], fold.test_idx[-1]))
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 3))
for row, fold in enumerate(folds):
    ax.scatter(
        fold.train_idx, [row] * len(fold.train_idx), s=4, label="train" if row == 0 else None
    )
    ax.scatter(fold.test_idx, [row] * len(fold.test_idx), s=8, label="test" if row == 0 else None)
    assert fold.train_idx[-1] < fold.test_idx[0]
ax.set(
    xlabel="Synthetic row index",
    ylabel="Fold",
    title="Expanding training windows and disjoint test folds",
)
ax.legend()
plt.show()
```

This plot checks split geometry, not model generalization. Set purge and embargo from label and information timing requirements.
