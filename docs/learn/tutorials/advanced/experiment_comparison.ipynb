# Matched experiment comparison

**Tier:** Docs-CI executable (CPU) · **Expected runtime:** under 10 seconds · **Hardware:** CPU

A controlled comparison keeps shared settings fixed and declares the intended change before examining outcomes. This notebook reads two catalog entries without launching training.

```python
from pathlib import Path

import yaml

ROOT = Path.cwd().resolve()
while not (ROOT / "pyproject.toml").is_file():
    ROOT = ROOT.parent

catalog = yaml.safe_load((ROOT / "config/experiments.yaml").read_text())
variants = {item["name"]: item for item in catalog["variants"]}
selected = list(variants.values())[:2]
print("selected variants:", [item["name"] for item in selected])
print("available fields:", sorted(set().union(*(item.keys() for item in selected))))
print("seed control belongs to the run request; no run is started here.")
```

```python
left, right = selected
left_settings = {key: value for key, value in left.items() if key not in {"name", "description"}}
right_settings = {key: value for key, value in right.items() if key not in {"name", "description"}}
changed = sorted(
    key
    for key in set(left_settings) | set(right_settings)
    if left_settings.get(key) != right_settings.get(key)
)
print("catalog-level differences:", changed)
print(
    "matched settings:",
    sorted(
        key
        for key in set(left_settings) & set(right_settings)
        if left_settings[key] == right_settings[key]
    ),
)
```
