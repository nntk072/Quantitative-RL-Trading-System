# Trade-setup policy schema

**Tier:** Docs-CI executable (CPU) · **Expected runtime:** under 10 seconds · **Hardware:** CPU

The setup action space maps a categorical selection to an idea, timeframe, and direction, plus execution coordinates. The no-trade choice is explicit.

```python
from collections import Counter

from decisioning.trade_setup import (
    NO_TRADE_ID,
    TRADE_SETUP_CATEGORY_COUNT,
    TRADE_SETUP_IDEAS,
    TRADE_SETUP_TIMEFRAMES,
    trade_setup_candidate,
    trade_setup_candidate_id,
    trade_setup_center,
)

print(
    "categories:",
    TRADE_SETUP_CATEGORY_COUNT,
    "ideas:",
    TRADE_SETUP_IDEAS,
    "timeframes:",
    TRADE_SETUP_TIMEFRAMES,
)
no_trade = trade_setup_candidate(NO_TRADE_ID)
po3_h1_long_id = trade_setup_candidate_id("po3", "H1", 1)
po3_h1_long = trade_setup_candidate(po3_h1_long_id)
print("no trade:", no_trade)
print("PO3 H1 long:", po3_h1_long, "box center:", round(trade_setup_center(po3_h1_long_id), 4))
assert TRADE_SETUP_CATEGORY_COUNT == 109
assert no_trade.is_no_trade
```

```python
counts = Counter(
    candidate.idea or "NO_TRADE"
    for candidate in map(trade_setup_candidate, range(TRADE_SETUP_CATEGORY_COUNT))
)
print(counts)
assert counts["NO_TRADE"] == 1
```

## Interpretation

The candidate category encodes intent and context. The environment still applies its eligibility mask and validates the resulting execution coordinates. Examples include PO3, order block, standard-deviation context, FVG retest, and IFVG retest.
