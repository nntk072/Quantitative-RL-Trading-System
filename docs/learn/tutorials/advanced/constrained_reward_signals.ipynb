# Reward and cost signals

A constrained objective tracks reward and a separate cost signal. The toy values below only illustrate a multiplier update; they are not a repository training callback or a research result. Hard environment risk limits remain separate.


```python
import matplotlib.pyplot as plt
import numpy as np

steps = np.arange(12)
reward = np.array([0.2, 0.3, -0.1, 0.4, 0.1, 0.2, -0.2, 0.35, 0.1, 0.25, -0.1, 0.3])
cost = np.array([0.02, 0.01, 0.05, 0.01, 0.03, 0.02, 0.06, 0.01, 0.02, 0.01, 0.04, 0.02])
limit, learning_rate, multiplier = 0.025, 0.5, 0.0
values = []
for c in cost:
    multiplier = max(0.0, multiplier + learning_rate * (c - limit))
    values.append(multiplier)
```

```python
fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)
axes[0].plot(steps, reward, marker="o")
axes[0].set_ylabel("Illustrative reward")
axes[1].plot(steps, cost, marker="o", label="cost")
axes[1].axhline(limit, color="red", linestyle="--", label="illustrative limit")
axes[1].set(xlabel="Step", ylabel="Illustrative cost")
axes[1].legend()
plt.tight_layout()
plt.show()
```

A learned penalty is not a substitute for hard risk enforcement. The numbers are arbitrary teaching values and support no economic or safety claim.
