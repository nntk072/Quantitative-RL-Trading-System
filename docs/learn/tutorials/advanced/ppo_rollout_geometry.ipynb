# PPO rollout geometry

This is an arithmetic visualization of PPO rollout size and optimizer minibatches. It does not build or train a policy.


```python
import math

import matplotlib.pyplot as plt

n_envs, n_steps, batch_size, epochs = 8, 256, 512, 4
samples = n_envs * n_steps
minibatches_per_epoch = math.ceil(samples / batch_size)
print(f"rollout samples: {samples:,}")
print(f"optimizer minibatches: {minibatches_per_epoch * epochs:,}")
assert samples == 2048
```

```python
fig, ax = plt.subplots(figsize=(8, 2.5))
ax.barh([0], [samples], color="#4477aa")
for x in range(0, samples, n_steps):
    ax.axvline(x, color="white", linewidth=1)
ax.set(
    yticks=[],
    xlabel="Transitions in one rollout",
    title=f"{n_envs} environments × {n_steps} steps = {samples} samples",
)
plt.show()
```

This calculation describes configured geometry, not the realized run. Check the run manifest for effective environment count and steps.
